```python
import pandas as pd, numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

P = '/content/'
day  = pd.read_csv(P + 'city_day_clean.csv', parse_dates=['Date'])
hour = pd.read_csv(P + 'city_hour_clean.csv')
day['City'] = day['City'].str.strip()
hour['City'] = hour['City'].str.strip()

print(day.shape, hour.shape)
print(day.columns.tolist())
print(hour.columns.tolist())
```

```python
dt_col = next(c for c in hour.columns if 'date' in c.lower())
hour[dt_col] = pd.to_datetime(hour[dt_col], errors='coerce')
hour['Date'] = hour[dt_col].dt.normalize()

keep = [c for c in ['PM2.5','PM10','NO2','CO','O3'] if c in hour.columns]
d = hour.groupby(['City','Date'])[keep].agg(['max','std'])
d.columns = [f'{a}_{b}' for a,b in d.columns]
d = d.reset_index()
HOUR_FEATS = [c for c in d.columns if c not in ('City','Date')]

df = day.merge(d, on=['City','Date'], how='left')
df['has_hourly'] = df[HOUR_FEATS[0]].notna().astype(int)

print(df.shape, '| rows with hourly match:', df['has_hourly'].sum())
print('cities with hourly data:', df.loc[df.has_hourly==1,'City'].nunique(), 'of', df.City.nunique())
```

```python
POLL = ['PM2.5','PM10','NO','NO2','NOx','NH3','CO','SO2','O3','Benzene','Toluene']

df['month']      = df['Date'].dt.month
df['weekday']    = df['Date'].dt.weekday
df['is_weekend'] = (df['weekday'] >= 5).astype(int)
df['is_winter']  = df['month'].isin([12,1,2]).astype(int)
df['season'] = df['month'].map(lambda m: 'Winter' if m in (12,1,2) else
                               'Summer' if m in (3,4,5) else
                               'Monsoon' if m in (6,7,8,9) else 'PostMonsoon')
diw = [c for c in df.columns if 'diwali' in c.lower()]
DIWALI = diw[0] if diw else None

train = df[df['Date'] <  '2019-01-01'].copy()
test  = df[df['Date'] >= '2019-01-01'].copy()
print(len(train), len(test))   # expect 13326 and 11475

def fill(train, test, cols):
    med = train.groupby(['City','month'])[cols].median()
    glob = train[cols].median()
    out = []
    for part in (train, test):
        part = part.copy()
        m = part.join(med, on=['City','month'], rsuffix='_med')
        for c in cols:
            part[c] = part[c].fillna(m[c + '_med']).fillna(glob[c])
        out.append(part)
    return out

train, test = fill(train, test, POLL + HOUR_FEATS)

BASE = POLL + ['month','weekday','is_weekend','is_winter'] + ([DIWALI] if DIWALI else [])
def X_of(part, extra):
    X = part[BASE + extra].copy()
    X = pd.concat([X, pd.get_dummies(part['season'], prefix='season'),
                      pd.get_dummies(part['City'],   prefix='city')], axis=1)
    return X

def run(extra, name):
    Xtr, Xte = X_of(train, extra), X_of(test, extra)
    Xte = Xte.reindex(columns=Xtr.columns, fill_value=0)
    rf = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
    rf.fit(Xtr, train['AQI'])
    p = rf.predict(Xte)
    return rf, p, Xtr.columns, {'Model': name,
        'MAE':  round(mean_absolute_error(test['AQI'], p),1),
        'RMSE': round(mean_squared_error(test['AQI'], p) ** 0.5,1),
        'R2':   round(r2_score(test['AQI'], p),3)}
```

```python
rf_a, p_a, cols_a, res_a = run([], 'A1: city_day features only')
print(res_a)
```

```python
rf_b, p_b, cols_b, res_b = run(HOUR_FEATS + ['has_hourly'], 'A2: + city_hour daily features')
print(res_b)

results = pd.DataFrame([res_a, res_b])
display(results)

# fair check: only test rows that actually have hourly data
mask = (test['has_hourly'] == 1).values
sub = pd.DataFrame([
    {'Model': 'A1 (hourly-covered rows)', 'MAE': round(mean_absolute_error(test['AQI'][mask], p_a[mask]),1),
     'R2': round(r2_score(test['AQI'][mask], p_a[mask]),3)},
    {'Model': 'A2 (hourly-covered rows)', 'MAE': round(mean_absolute_error(test['AQI'][mask], p_b[mask]),1),
     'R2': round(r2_score(test['AQI'][mask], p_b[mask]),3)}])
display(sub)
```

```python
import matplotlib.pyplot as plt
imp = pd.Series(rf_b.feature_importances_, index=cols_b).sort_values().tail(15)
imp.plot.barh(figsize=(7,5), title='Model A2: top 15 features')
plt.tight_layout()
plt.savefig(P + 'model_a_importance.png', dpi=150)
plt.show()

results.to_csv(P + 'model_a_results.csv', index=False)
```
